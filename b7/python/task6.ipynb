## Semiconductor Lab B7 - Task 6

Estimate the power density in the active volume in the laser mode from the given
geometric values of the laser.

```python
from pprint import pprint
```

### Import Data


`V_op`, `I_op` and `eta_total` are copied from the P-I table in `task1.ipynb` (the row at
19 mA) rather than re-read from the raw data, since this task only needs the one operating
point.

```python
V_op: float = 1.810  # V, task1.ipynb P-I table at I_mA = 19.00
I_op: float = 0.019  # A
eta_total: float = 0.026418  # P_opt / P_el at this operating point, task1.ipynb

P_el: float = V_op * I_op
P_opt: float = eta_total * P_el

pprint({"P_el_W": P_el, "P_opt_W": P_opt})
```

### Enrich Data

`L_NM`, `W_NM` and `D_NM` are the stripe length, the stripe width, and the
active-layer thickness 

```python
L_NM: float = 300_000.0  # nm
W_NM: float = 5_000.0  # nm
D_NM: float = 20.0  # nm

V_active_m3: float = (L_NM * W_NM * D_NM) * 1e-27  # nm^3 -> m^3

pprint({"V_active_m3": V_active_m3, "V_active_cm3": V_active_m3 * 1e6})
```

### Result

```python
power_density_W_m3: float = P_opt / V_active_m3
power_density_W_cm3: float = power_density_W_m3 * 1e-6

pprint({
    "power_density_W_per_m3": f"{power_density_W_m3:.2e}",
    "power_density_W_per_cm3": f"{power_density_W_cm3:.2e}",
})
```

- At 19mA, the registered optical power is 0.9mW
- By estimating the active volume as 

L=300000nm    W = 5000 nm D=20 nM -> V = 3e-17 m^3

the power density can be calculated as P_opt / V = $3.0 \times 10^{13}\,\mathrm{W\,m^{-3}}$ ($3.0 \times 10^{7}\,\mathrm{W\,cm^{-3}}$)
